## Square Root Problem (Newton-Raphson / Babylonian Method)
The general principle is that if two numbers multiply together to give the target number $x$ (for which we want to find the square root), then unless they are exactly the square root, one must be too small and the other must be too large.

The true square root lies somewhere between these two numbers. By calculating their average, we can find a better approximation.

Specifically, for a guess $g$:
- If $g$ is too small (i.e., $g < \sqrt{x}$), then $x / g$ is too large ($x / g > \sqrt{x}$).
- If $g$ is too large (i.e., $g > \sqrt{x}$), then $x / g$ is too small ($x / g < \sqrt{x}$).

In either case, the actual square root lies between $g$ and $x / g$.
By repeatedly averaging $g$ and $x / g$, we will eventually converge toward the correct answer.


````python
def newton_sqrt(x: float, tol: float = 1e-10) -> float:
    """Compute the square root of x using Newton's method."""
    if x < 0:
        raise ValueError("Cannot compute square root of negative number.")
    if x == 0:
        return 0.0
    guess = x / 2.0
    while True:
        next_guess = (guess + x / guess) / 2.0
        if abs(next_guess - guess) < tol:
            return next_guess
        guess = next_guess
````

### Scientific Notation

In Python, scientific notation is written using the letter `e` or `E` to represent "times ten to the power of".

For example:
- `1e-10` means $1 \times 10^{-10}$ (or $0.0000000001$)
- This is used as the default tolerance (`tol`) in our square root calculations to define how close a guess needs to be before we consider it "equal".

### Floats
Floating-point numbers are not the same as mathematical real numbers; they are represented in 64 bits, providing about 15–17 decimal digits of precision.

Due to floating-point representation errors, we should check if values are "close enough" rather than exactly "equal" — which is why we use a tolerance (`tol`) parameter.


### Caesar Cipher
In a Caesar cipher, every letter in the plaintext is shifted by a fixed number of positions down the alphabet.

**How to implement/solve this in Python?**
```python
>>> ALPHABET = "abcdefghijklmnopqrstuvwxyz"
>>> ALPHABET[3]
'd'
>>> ALPHABET.index('d')
3
```

- Shifting `'a'` by 3: position $0 + 3 = 3$, which is `'d'`.
- Shifting `'z'` by 3: position $25 + 3 = 28$. Since there is no index 28, we wrap around using modulo arithmetic:
```python
>>> 28 % 26
2
```
Position 2 corresponds to `'c'`.


````python
def caesar_cipher(text: str, shift: int) -> str:
    """Encrypt text using a Caesar cipher with the given shift.
    
    >>> caesar_cipher("abc", 3)
    'def'
    >>> caesar_cipher("xyz", 3)
    'abc'
    >>> caesar_cipher("Hello, World!", 5)
    'Mjqqt, Btwqi!'
    """
    result = []
    for char in text:
        if char.isalpha():
            shift_base = ord('A') if char.isupper() else ord('a')
            result.append(chr((ord(char) - shift_base + shift) % 26 + shift_base))
        else:
            result.append(char)
    return "".join(result)
````

### Lists
- `[]` is an empty list. `result.append(x)` adds `x` to the end.
- A string cannot be changed in-place (strings are immutable). So we build a list of characters/substrings, then join them.
- `"".join(result)` glues a list of strings into one single string, with `""` as the delimiter.
- `for c in s:` walks through a string one character at a time.
- `c in ALPHABET` asks a question: is the character present in the alphabet? Notice that `in` has two slightly distinct uses in Python: iterating over a sequence (`for c in s`) and membership testing (`c in ALPHABET`).


### Slicing
Slicing allows you to extract portions of sequences (strings, lists, etc.) using the `[start:stop:step]` syntax:
- `s[2:5]`: from index 2, up to but not including 5.
- `s[2:]`: from index 2 to the end of the sequence.
- `s[:5]`: the first 5 elements (indices 0 to 4).
- `s[-1]`: the last item.
- `s[-3:]`: the last 3 items.
- `s[::-1]`: the entire sequence in reverse.
- Slicing a string returns a new string; slicing a list returns a new list.
- Slicing always returns a new object (shallow copy).


### Tuples
`(k, text)` is a tuple.
- Tuples use parentheses `( )` and are immutable (once created, they cannot be modified).
- You can unpack them in a `for` loop: `for k, text in candidates:`.
- **Best Practice:**
  - Use a **tuple** when the elements represent a heterogeneous structure (the parts have different meanings/types, e.g., an ID and a message).
  - Use a **list** when the items are homogeneous (the same kind of data, e.g., a collection of strings or numbers) and can dynamically grow or shrink.
